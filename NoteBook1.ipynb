```python
Products["category"] = Products["category"].str.strip()
Products["name"] = Products["name"].str.strip()

# Standardize category names
Products["category"] = Products["category"].str.title()
```

```python
Products["product_id"] = Products["product_id"].astype(int)

Products["price_usd"] = pd.to_numeric(
    Products["price_usd"],
    errors="coerce"
)

Products["cost_usd"] = pd.to_numeric(
    Products["cost_usd"],
    errors="coerce"
)
Products["margin_usd"] = pd.to_numeric(
    Products["margin_usd"],
    errors="coerce"
)
```

```python
Products["calculated_margin"] = (
    Products["price_usd"] -
    Products["cost_usd"]
)

margin_errors = Products[
    abs(
        Products["margin_usd"] -
        Products["calculated_margin"]
    ) > 0.01
]

print("Margin Errors:", len(margin_errors))
```
